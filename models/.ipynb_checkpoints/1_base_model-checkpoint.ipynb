```python
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset, random_split
from tqdm import tqdm
```

# Introduction
iVAE was introduced in  the 2020 paper "Variational autoencoders and nonlinear ICA: A unifying framework" by Khemakhem et al. It assumes a factorized distribution $$p(x, z, u) = p(u)p(z|u)p(x|z),$$ where x is observed data, u is an auxiliary label, and z is the latent space.

The original iVAE model is reproduced in "Model definitions". The four main metrics of evaluation are then introduced. The model is trained using the original synthetic data from iVAE, and then evaluated on each of the metrics on test data.

# Model definitions

### Utility functions

```python
def sample(mu, logvar):
    z = mu + (0.5 * logvar).exp() * torch.randn_like(mu)
    return z

# Log likelihood of multivariate normal distribution with diagonal covariance
def normal_ll(x, mu, logvar):
    const = -x.shape[-1] * torch.tensor(2*torch.pi).log()
    ll = -0.5 * (const + logvar + (x - mu).pow(2) / logvar.exp()).sum(-1)
    return ll
```

### MLP network

```python
class MLP(nn.Module):
    def __init__(self, in_dim, out_dim, mu_flag=True, var_flag=True):
        super().__init__()
        self.mu_flag = mu_flag    # mu may be learned or fixed to a constant
        self.var_flag = var_flag  # logvar may be learned or fixed to a constant
        hidden = 50

        self.layers = nn.Sequential(
            nn.Linear(in_dim, hidden),
            nn.LeakyReLU(1e-2),    # LeakyReLU is used to facilitate the injective mapping from Z to X
            nn.Linear(hidden, hidden),
            nn.LeakyReLU(1e-2)
        )
        
        if self.mu_flag:
            self.mu = nn.Linear(hidden, out_dim)
        if self.var_flag:
            self.logvar = nn.Linear(hidden, out_dim)

    def forward(self, x):
        out = self.layers(x)
        if self.mu_flag and self.var_flag:
            return self.mu(out), self.logvar(out)
        if self.mu_flag:
            return self.mu(out)
        if self.var_flag:
            return self.logvar(out)
```

### iVAE model

```python
class iVAE_Base(nn.Module):
    def __init__(self, kx, ku, **kwargs):
        super().__init__()

        pz = 2  # dimensionality of latent space

        # Prior p(z|u)
        self.mu_prior = torch.zeros(pz)  # prior mean fixed to zero
        self.prior_logvar = MLP(ku, pz, mu_flag=False)

        # Encoder q(z|x,u)
        self.encoder_mu = MLP(kx + ku, pz, var_flag=False)
        self.encoder_logvar = MLP(kx + ku, pz, mu_flag=False)
        
        # Decoder p(x|z)
        self.decoder_mu = MLP(pz, kx, var_flag=False)
        noise = torch.tensor(0.01)
        self.logvar_p = noise.log() * torch.ones(kx)  # decoder covariance fixed to 0.01

    def forward(self, x, u):
        xu = torch.cat((x, u), 1)
        # Prior parameters
        self.logvar_prior = self.prior_logvar(u)

        # Encoder parameters
        self.mu_q = self.encoder_mu(xu)
        self.logvar_q = self.encoder_logvar(xu)

        # Reparameterization trick
        z = sample(self.mu_q, self.logvar_q)
        self.z = z

        # Decoder parameters
        self.mu_p = self.decoder_mu(z)

        return self.mu_p, self.logvar_p, self.mu_prior, self.logvar_prior, self.mu_q, self.logvar_q

    def loss(self, x):
        mu_prior = self.mu_prior.expand(x.shape[0], -1)
        px_z = normal_ll(x, self.mu_p, self.logvar_p)          # log p(x|z)
        qz_xu = normal_ll(self.z, self.mu_q, self.logvar_q)    # log p(z|x,u)
        pz_u = normal_ll(self.z, mu_prior, self.logvar_prior)  # log p(z|u)
        return torch.mean(px_z), torch.mean(qz_xu), torch.mean(pz_u)

    def get_z_samples(self):
        return self.z
```

# Metrics of evaluation

### Active units
This measures "useful" latents by how much variance the latents have to different inputs. Low variance is an indicator of posterior collapse, i.e. the latents do not depend on the data.
From "Importance weighted autoencoder", Burda et al, 2016.

```python
def active_units(mu_q, thresh=1e-3):
    au_var = mu_q.var(0)
    au = au_var > thresh
    return au.detach().cpu().numpy(), au_var.detach().cpu().numpy()
```

### Approximate mutual information

Estimates the approximate mutual information I(X;Z) under joint distribution q(X,Z) = p(X) q(Z|X). Used in "Posterior collapse and latent variable non-identifiability", Wang et al, 2021.

```python
# KL divergence between two multivariate normal distributions with diagonal covariance
# See: https://en.wikipedia.org/wiki/Multivariate_normal_distribution#Kullback%E2%80%93Leibler_divergence
def kl_normal(mu0, log_var0, mu1, log_var1):
    dim = mu0.shape[1]
    trace = log_var0.exp() / log_var1.exp()
    quad = (mu0 - mu1).pow(2) / log_var1.exp()
    logdet = log_var1 - log_var0
    kl_div = (trace + quad + logdet).sum(1) - dim
    return 0.5*kl_div

def approximate_mi_old(mu_q, logvar_q):
    batch_size, dim = mu_q.shape

    # Compute q(z) = N(mu_c,logvar_c) FIXME: wtf???
    mu_c = mu_q.sum(0)
    var_c = logvar_q.exp().sum(0)
    logvar_c = var_c.log()

    # Compute KL[q(z|xi) || q(z)] for each sample i
    mu_c = mu_c.unsqueeze(0).expand(batch_size, dim)
    logvar_c = logvar_c.unsqueeze(0).expand(batch_size, dim)
    kls = kl_normal(mu_q, logvar_q, mu_c, logvar_c)
    ami = kls.mean().detach().cpu().numpy()

    return ami


def approximate_mi(mu, logvar, z):
    # I_q(X;Z) = E_x[ q(z|x) log q(z|x)/q(z)] = E_x[q(z|x) (log q(z|x) - log q(z))]
    # log q(z) = -log|X| + log sum_x q(z|x)  <-- employ logsumexp trick here, |X| is batch size

    # Set up
    batch_size, dim = mu.shape
    z0 = z - mu
    
    # Calculate terms for: log q(z|x) - log q(z)
    log_post = dim*np.log(2*np.pi) + logvar + z0.pow(2)*logvar.exp().reciprocal()
    log_post = -0.5 * log_post.sum(1)
    posterior = log_post.exp()
    aggreg_post = torch.logsumexp(log_post, dim=0) - np.log(batch_size)

    # Calculate: q(z|x) (log q(z|x) - log q(z))
    inside_exp = posterior*(log_post - aggreg_post)

    # Calculate: E_x[q(z|x) (log q(z|x) - log q(z))]
    approx_mi = inside_exp.mean().item()
    return approx_mi


def approximate_mi_aux(mu, logvar, z, u):
    # I_p(M;Z) = E_m[ sum_z p(z|m) log p(z|m)/p(m)] = E_m[ sum_z q(z|m) (log p(z|m) - log p(m))]

    # Set up
    batch_size, dim = mu.shape
    z0 = z - mu

    one_idx = u.nonzero()[:, 1]  # NOTE: assumes one-hot encoding for message var
    unique_u = one_idx.unique()
    mi_per_u = torch.zeros(len(unique_u))

    for i, mval in enumerate(unique_u):
        # Select all z samples from a given m
        all_i = (one_idx == mval).nonzero().squeeze(dim=1)
        zi = z0[all_i, :]
        logvar_qi = logvar[all_i, :]

        # Calculate terms for: log p(z|m) - log p(z)
        log_post = dim * np.log(2 * np.pi) + logvar_qi + zi.pow(2) * logvar_qi.exp().reciprocal()
        log_post = -0.5 * log_post.sum(1)
        posterior = log_post.exp()
        aggreg_post = torch.logsumexp(log_post, dim=0) - np.log(batch_size)

        # Calculate: p(z|m) (log p(z|m) - log p(z))
        inside_exp = posterior * (log_post - aggreg_post)

        # Calculate: sum_z p(z|m) (log p(z|m) - log p(z))
        mi_per_u[i] = inside_exp.sum().item()

    # Calculate: E_m[ sum_z p(z|m) (log p(z|m) - log p(z)) ]
    approx_mi = mi_per_u.mean().item()
    return approx_mi
    
```

### Mean correlation coefficient
Measures how close estimated latents are to ground truth latents by Pearson correlation between z_true and z_hat (the closer to 1, the better). Provides optimal latent dimension matching via linear sum assignment. From "Variational autoencoders and nonlinear ICA: A unifying framework", Khemakhem et al, 2020.

```python
from scipy.optimize import linear_sum_assignment

def mcc(z_true, mu_q, logvar_q):
    dim = mu_q.shape[1]
    z_hat = mu_q + torch.exp(0.5*logvar_q) * torch.randn_like(mu_q)
    z_hat = z_hat.detach().cpu().numpy()
    z_true = z_true.detach().cpu().numpy()
    mcc_mat = np.zeros((dim, dim))
    for i in range(dim):
        for j in range(dim):
            mcc_mat[i, j] = np.corrcoef(z_true[:, j], z_hat[:, i])[0, 1]

    row_idx, col_idx = linear_sum_assignment(1-abs(mcc_mat))
    mapping = list(col_idx)
    return mcc_mat, mapping
```

### Decoding accuracy
Measures how well auxiliary label U can be decoded from observed data X by estimating p(u|x). From "Learning identifiable and interpretable latent models of high-dimensional neural activity using pi-VAE", Zhou and Wei, 2020.

```python
def decode_lbl(model, x, u_events, N=100):
    batch_size, kx = x.shape
    ll_all = []

    for val in range(u_events):
        # Create copies of the specific m value across all samples
        m = torch.zeros(u_events)
        m[val] = 1  # make one-hot encoding
        m = m.expand(batch_size, u_events)
        mu_prior = model.mu_prior
        logvar_prior = model.prior_logvar(m)

        # Draw N samples from p(z|m)
        mu_prior = mu_prior.unsqueeze(0).expand(N, -1, -1)
        logvar_prior = logvar_prior.unsqueeze(0).expand(N, -1, -1)
        z = mu_prior + (0.5*logvar_prior).exp() * torch.randn(N, batch_size, 2)

        # Input these z samples into the decoder to get mu_p, sigma_p
        z_flat = z.reshape(N*batch_size, -1)
        mu_p = model.decoder_mu(z_flat).reshape(N, batch_size, -1)
        logvar_p = model.logvar_p #model.decoder_logvar(z_flat).reshape(N, batch_size, -1)

        # Calculate log likelihood with each (mu_p, sigma_p) pair
        x_exp = x.unsqueeze(0)
        ll = normal_ll(x_exp, mu_p, logvar_p)
        ll_total = torch.logsumexp(ll, dim=0) - torch.log(torch.tensor(N, dtype=ll.dtype))
        ll_all.append(ll_total)#.detach().cpu().numpy())
        
    logliks = torch.stack(ll_all, dim=0).transpose(1, 0)
    log_p_m_given_x = torch.nn.functional.log_softmax(logliks, dim=0)
    predictions = log_p_m_given_x.argmax(1)
    return predictions, logliks
```

# Training

### Prepare for training

```python
rng = 1
torch.manual_seed(rng)
np.random.seed(rng)
```

```python
# Load raw data
data = np.load("../../data/data_ivae1.npz")
x = data['x']
u = data['u']
z = data['s']

# Get dimensions of data and auxiliary label
kx = x.shape[1]
ku = u.shape[1]

# Prepare dataset
batch_size = 64
data = TensorDataset(torch.Tensor(x), torch.Tensor(u), torch.Tensor(z))
generator = torch.Generator().manual_seed(rng)
train_data, test_data = random_split(data, [0.8, 0.2], generator=generator)
train_loader = DataLoader(train_data, batch_size, shuffle=True)
```

```python
# Initialize structures for saving meta data
recon_arr = []
kl_arr = []
```

```python
# Define model
model = iVAE_Base(kx, ku)
optimizer = torch.optim.Adam(params=model.parameters(), lr=1e-3)
# NOTE: original iVAE includes a scheduler. It makes training smoother but slower with the same outcome, so we omit it here
# scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, factor=0.1, patience=4)
```

### Main training loop

```python
epochs = 10

#train(model, optimizer, train_loader)

for epoch in tqdm(range(epochs)):

    for batch in train_loader:
        x = batch[0]
        u = batch[1]
        output = model(x, u)

        optimizer.zero_grad()
        losses = model.loss(x)
        recon_loss = -losses[0]
        kl_loss = losses[1] - losses[2]
        
        loss = recon_loss + kl_loss
        loss.backward()
        optimizer.step()

        recon_arr.append(recon_loss.item())
        kl_arr.append(kl_loss.item())
```
